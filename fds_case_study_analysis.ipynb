## Part I: Retail Transaction Clustering

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN, SpectralClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Load dataset and convert to binary matrix
df_retail = pd.read_csv('retail_transactions.csv', header=None)
print(f"Loading retail dataset... Shape: {df_retail.shape}")

binary_matrix = df_retail.notnull().astype(int)
print(f"Binary basket matrix shape: {binary_matrix.shape}")
```

```python
# Apply PCA for feature reduction
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(binary_matrix)
explained_variance = pca.explained_variance_ratio_ * 100
print(f"PCA applied. Explained variance - PC1: {explained_variance[0]:.2f}%, PC2: {explained_variance[1]:.2f}% (Total: {sum(explained_variance):.2f}%)")
```

```python
# Fit clustering algorithms
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_pca)

hierarchical = AgglomerativeClustering(n_clusters=4)
hier_labels = hierarchical.fit_predict(X_pca)

dbscan = DBSCAN(eps=0.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_pca)

spectral = SpectralClustering(n_clusters=4, random_state=42, gamma=1.0)
spectral_labels = spectral.fit_predict(X_pca)

print("Clustering execution completed successfully!")
```

## Part II: Predicting Student Dropout and Academic Success

```python
# Load student dataset and check distribution
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

df_students = pd.read_csv('student_dataset.csv')
print(f"Loading dataset... Shape: {df_students.shape}")

print("\nTarget class distribution:")
print(df_students['Target'].value_counts())
```

```python
# Train-test split
X = df_students.drop(columns=['Target'])
y = df_students['Target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape (n = {len(y_test)}): {X_test.shape}")
```

```python
# Train Logistic Regression model
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Model training and prediction completed!")
```

```python
# Model evaluation
print("Classification Report:")
print(classification_report(y_test, y_pred))
accuracy = accuracy_score(y_test, y_pred)
print(f"Overall Accuracy: {accuracy:.4f}")
```
